# Funções de Ingestão

```python
# Lista de datas de referência para atualização do Feature Store
dates = ['2018-10', '2018-11', '2018-12', '2019-01', '2019-02', '2019-03', '2019-04', '2019-05', '2019-06', '2019-07', '2019-08', '2019-09', '2019-10', '2019-11', '2019-12', '2020-01', '2020-02', '2020-03', '2020-04', '2020-05', '2020-06', '2020-07', '2020-08', '2020-09', '2020-10', '2020-11', '2020-12', '2021-01', '2021-02', '2021-03', '2021-04', '2021-05', '2021-06']
```

```python
def update_feature_store(catalog, database, table, query_file, dates):
    tablename = f"{catalog}.{database}.{table}"

    # Lê o arquivo SQL contendo a query parametrizada
    with open(query_file, "r") as f:
        query = f.read()

    fe = FeatureEngineeringClient()

    # Verifica se a tabela já existe no Feature Store
    def table_exist(catalog, database, table):
        return spark.sql(
            f"SHOW TABLES FROM {catalog}.{database} LIKE '{table}'"
        ).count() > 0

    if table_exist(catalog, database, table):
        # Atualiza a tabela existente para cada data de referência
        for dt_ref in dates:
            print(f"Escrevendo {dt_ref}")
            df = spark.sql(
                query.format(dt_ref=f"{dt_ref}-01")
            )
            fe.write_table(
                name=tablename,
                df=df,
                mode="merge"
            )
        print("Table updated")
    else:
        # Cria a tabela no Feature Store com a primeira data
        first_date = dates[0]
        df = spark.sql(
            query.format(dt_ref=f"{first_date}-01")
        )
        fe.create_table(
            name=tablename,
            primary_keys=["ID_CLIENTE", "ID_DOCUMENTO", "DATA_REF"],
            partition_columns=["DATA_REF"],
            df=df
        )
        print("Table created")
        # Escreve as demais datas na tabela recém-criada
        for dt_ref in dates[1:]:
            print(f"Escrevendo {dt_ref}")
            df = spark.sql(
                query.format(dt_ref=f"{dt_ref}-01")
            )
            fe.write_table(
                name=tablename,
                df=df,
                mode="merge"
            )
        print("Table created")
```

# Ingestão de Dados

## Cadastral

```python
update_feature_store(
    catalog="feature_store",
    database="credit_score",
    table="fs_cadastral",
    query_file="fs_cadastral.sql",
    dates=dates
)
```

# Ingestão Final Antes da Previsão

```python
# Data de referência para ingestão final de features para previsão
dates = ['2021-07']

# Lista de tabelas e arquivos SQL correspondentes
tables_queries = [
    ("fs_cadastral", "fs_cadastral.sql"),
    ("fs_temporal", "fs_temporal.sql"),
    ("fs_historico_financeiro", "fs_historico_financeiro.sql"),
    ("fs_renda", "fs_renda.sql"),
    ("fs_funcionarios", "fs_funcionarios.sql"),
    ("fs_historico_pagamentos", "fs_historico_pagamentos.sql")
]

# Ingestão das features finais no Feature Store para previsão
for table, query_file in tables_queries:
    update_feature_store(
        catalog="feature_store",
        database="credit_score",
        table=table,
        query_file=query_file,
        dates=dates
    )
```
